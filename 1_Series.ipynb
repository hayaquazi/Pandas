Pandas Series- a seriesi is a one dimensional labeled array capable of holding any data type. The axis labels are colletively called the index

```python
import numpy as np
import pandas as pd
```

```python
labels =['a','b','c'] #index
my_list = [10,20,30]
arr = np.array([10,20,30])
d = {'p':10, 'q':20, 'r':30}
```

```python
pd.Series(my_list)
```

```python
pd.Series(my_list, index = labels)
```

```python
pd.Series(arr) #only one dimen
```

```python
pd.Series(d)
```
