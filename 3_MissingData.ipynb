Finding Missing Data - Based on Column

```python
import numpy as np
import pandas as pd
```

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
df.isna() #Checking null values 
```

```python
df.isna().sum() #Total null values
```

```python
df.isna().any() #checking is there any null value in any column
```

Removing Null value - Work on the basis of Rows


```python
df.dropna()  #drop entire row containing atleast one null value
```

```python
df
```

```python
df.dropna(thresh = 3) # atleast 3 non nll vlaue
```

```python
df.dropna(thresh = 1)
```

Filling Null data - Based Column 

```python
df.fillna(0) #temporary
```

```python
#filling with different values
values ={'A':0, 'B':100,'C':300, 'D':400}
df.fillna(value = values)
```

```python
df.fillna(df.mean())
```
